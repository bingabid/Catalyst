```python
import os
import numpy as np
import seaborn as sns
from configs import ModelConfig
import matplotlib.pyplot as plt
```

```python
model = 'resnet18'
in_dataset = 'CIFAR-10'
# model = 'resnet34'
# in_dataset = 'CIFAR-10'
# model = 'densenet101'
# in_dataset = 'CIFAR-10'

# model = 'resnet_imagenet34'
# in_dataset = 'ImageNet-1K'
```

```python
base_directory_name = f"{model}/{in_dataset}/"
model_checkpoint_directory = os.path.join(ModelConfig.model_checkpoint_directory, base_directory_name)
model_statistics_directory = os.path.join(ModelConfig.model_statistics_directory, base_directory_name)
print(model_statistics_directory)
in_model_statistics_directory = os.path.join(model_statistics_directory, f"{in_dataset}")
in_statistics_file_name = os.path.join(in_model_statistics_directory, f"in_{in_dataset}")
print(in_statistics_file_name)
W_fname = in_statistics_file_name + '_W.npy'
b_fname = in_statistics_file_name + '_b.npy'
print(W_fname)
print(b_fname)
W = np.load(W_fname)
b = np.load(b_fname)
W.shape, b.shape
```

```python
plt.figure(figsize=(10, 6))
sns.heatmap(W, cmap='viridis', cbar=True, xticklabels=False, yticklabels=False)
plt.title("Heatmap of Weight Matrix")
plt.xlabel("Units")
plt.ylabel("Classes")
plt.tight_layout()
plt.show()
```

```python
pos_count = (W >0.0).sum(axis=1)
print(pos_count)
neg_count = (W <0.0).sum(axis=1)
print(neg_count)
```

```python
pos_count = (W >0.2).sum(axis=1)
print(pos_count)
neg_count = (W <-0.2).sum(axis=1)
print(neg_count)
```
